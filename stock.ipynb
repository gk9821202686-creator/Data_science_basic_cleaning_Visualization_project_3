```python
import pandas as pd
import numpy as np
df = pd.read_csv('stocks.csv')
df["date"] = pd.to_datetime(df["Date"],format="%m/%d/%Y").dt.strftime("%d").copy()
org = df[["Ticker","date","High","Low","Open","Close"]]
# isnull used to check any value is missing or not
org
```

```python
# seperate all company data and remove all missing values
a = org.query("Ticker == 'AAPL'").copy()
apple = a.dropna(axis=0).reset_index(drop=True).copy().head(20) #reset index and copy the data in apple
apple = apple.assign(profit_loss = apple.Close - apple.Open).copy()
a
```

```python
b = org.query("Ticker == 'MSFT'").copy()
microsoft = b.dropna(axis=0).reset_index(drop=True).copy().head(20)
microsoft = microsoft.assign(profit_loss =  microsoft.Close - microsoft.Open).copy()
b
```

```python
c = org.query("Ticker=='GOOG'").copy()
google = c.dropna(axis=0).reset_index(drop=True).copy().head(20)
google = google.assign(profit_loss = google.Close - google.Open).copy()
c
```

```python
d = org.query("Ticker=='NFLX'").copy()
netflix = d.reset_index(drop=True).head(20).copy()
netflix = netflix.assign(profit_loss = netflix.Close - netflix.Open).copy()
d
```

```python
import matplotlib.pyplot as plt
apple.plot(kind="bar",x="date",y="profit_loss",color='black')
plt.title("Apple stock profit loss")
plt.xlabel("30 days feb-mar")
plt.ylabel("Price")
plt.grid()

microsoft.plot(kind = "bar",x="date",y="profit_loss",color = 'blue')
plt.title("Microsoft stock profit loss")
plt.xlabel("30 day from Feb-march")
plt.ylabel("Price")
plt.grid()

google.plot(kind="bar",x="date",y="profit_loss",color = 'g')
plt.title("Google stock profit loss")
plt.xlabel("30 day from feb-march")
plt.ylabel("Price")
plt.grid()

netflix.plot(kind="bar",x="date",y ="profit_loss",title="profit and loss",color = "r")
plt.title("Netflix stock profit loss")
plt.xlabel("30 day from feb-mar")
plt.ylabel("price")
plt.grid()


```
